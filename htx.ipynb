### **System Requirements Estimation for a 50,000-Camera Deployment**

We estimate the bandwidth and storage requirements for a video surveillance system comprising 50,000 cameras, each streaming video at 720p resolution, 30 FPS, compressed using H.264, with an average bitrate of 3 Mbps per camera.

---
* $N = 50{,}000$ be the number of cameras
* $B_c = 3 \, \text{Mbps}$ be the average bitrate per camera
* $B_{total}$ be the total bandwidth requirement


$$
B_{total} = N \times B_c = 50{,}000 \times 3 \, \text{Mbps} = 150{,}000 \, \text{Mbps}
$$

$$
B_{total} = \frac{150{,}000}{1000} \, \text{Gbps} = 150 \, \text{Gbps}
$$

$$
B_{total} = \frac{150 \times 10^9}{8} \, \text{Bytes/s} = 18.75 \, \text{GB/s}
$$

This represents the total ingress bandwidth at the on-prem HTX's Video Management System (VMS).

---

#### **Storage Requirements**

* $S_c$ be the storage per camera per hour
* $T = 3600 \, \text{s/hour}$
* $R_c = \frac{B_c}{8} = 0.375 \, \text{MB/s}$ disk size per camera

$$
S_c = R_c \times T = 0.375 \, \text{MB/s} \times 3600 \, \text{s} = 1350 \, \text{MB} = 1.35 \, \text{GB/hour}
$$

Daily storage per camera:

$$
S_{c,day} = 1.35 \, \text{GB/hour} \times 24 \, \text{hours} = 32.4 \, \text{GB/day}
$$

Daily storage for all 50K cameras:

$$
S_{total,day} = N \times S_{c,day} = 50{,}000 \times 32.4 \, \text{GB} = 1{,}620{,}000 \, \text{GB} = 1.62 \, \text{PB/day}
$$

Monthly storage (30 days):

$$
S_{total,month} = 1.62 \, \text{PB/day} \times 30 = 48.6 \, \text{PB/month}
$$

⚠️ Note: These are compressed with H.264 Compression and no redundancy is considered
